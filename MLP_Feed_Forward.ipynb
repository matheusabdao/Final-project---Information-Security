### X = 3 inputs [1, 2, 3]
### y = output vector of 2 neurons
### W = weight matrix between input and the first hidden layer
### v = weight matrix between the first hidden layer and ouput hidden layer has 4 neurons
### implement a MLP with one hidden layer
### how many hyper-parameter we have in this network?

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Num of inputs, hidden and outputs

```python
n_inputs = 3
n_hidden = 4
n_outputs = 2
```

# Weight

```python
W_inp_hid = np.random.rand(n_inputs, n_hidden)
w_hid_out = np.random.rand(n_hidden, n_outputs)
```

# Define the activation function for the hidden layer

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

# Calculate the output of the hidden layer/output layer

```python
def forward_pass(X):
    hidden_layer = sigmoid(np.dot(X, W_inp_hid))
    output_layer = sigmoid(np.dot(hidden_layer, w_hid_out))
    return output_layer
```

# Test the network

```python
X = np.array([1, 2, 3])
output = forward_pass(X) # output = sigmoid(sigmoid(X * W_inp_hid) * w_hid_out)
print("Input: \n",X)
print("\nInput to hidden layer: \n",W_inp_hid)
print("\nHidden layer to output: \n",w_hid_out)
print("\nOutput: \n", output)
```

```python
x_values = np.linspace(-10, 10, 100)
y_sigmoid = sigmoid(x_values)
```

```python
plt.plot(x_values, y_sigmoid, label='Sigmoid Function', color='blue')

plt.scatter(np.arange(len(output)), output, color='red', label='Network Outputs')

plt.title('Sigmoid Function and MLP Outputs')
plt.xlabel('X / Neuron Index')
plt.ylabel('Sigmoid / Output')
plt.legend()
plt.grid(True)
plt.show()
```

```python
hyper_parameters = n_inputs * n_hidden + n_hidden * n_outputs
print("Number of  hyper-parameters in this network:", hyper_parameters)
```

# As for the number of hyper-parameters, we have:


The weights in the weight matrix W (n_inputs * n_hidden) = 12 hyper-parameters

The weights in the weight matrix v (n_hidden * n_outputs) = 8 hyper-parameters

So, in total, we have 12 + 8 = 20 hyper-parameters in this network.
